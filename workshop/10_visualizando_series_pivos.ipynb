### Importando bibliotecas

### Seleção da área de interesse

```python
# Seleção do pivô
#id_pivo = 97 # TODO: Defina o código do pivô de interesse
id_pivo = 'B'
title = f'ID Pivô: {id_pivo}'

# Pasta com os resultados de séries temporais para uma seleção de pivos centrais
folder_path = r"E:\Workshop_ANA_2026\SerieTemporal_Pivos_ptjpl_CRA" #PT-JPL
folder_path= r"E:\Workshop_ANA_2026\SerieTemporal_Pivos_sims_CRA" #SIMS

# Lista de arquivos na pasta
csv_files = [f for f in os.listdir(folder_path) if f.endswith('.csv')]
print(csv_files)
```

### Lendo resultados

```python
# Gráficos para um pivo selecionado
id_pivo = 'B' #TODO: Defina o código do pivô de interesse
file = f'SerieTemporal_Pivos_ptjpl_CRA_{id_pivo}.csv'
title = f'ID Pivô: {id_pivo}'

id_pivo = 'B' #TODO: Defina o código do pivô de interesse
file = f'SerieTemporal_Pivos_sims_CRA_{id_pivo}.csv'
title = f'ID Pivô: {id_pivo}'

# Gráfico de ET média no pivo

# lê a série PT-JPL
ptjpl = pd.read_csv(r"E:\Workshop_ANA_2026\SerieTemporal_Pivos_ptjpl_CRA\ptjpl_pivo_B.csv", index_col='date')
ptjpl.index = pd.to_datetime(ptjpl.index)
ptjpl = ptjpl.dropna()
ptjpl.sort_index(inplace=True)

# lê a série SIMS
sims = pd.read_csv(r"E:\Workshop_ANA_2026\SerieTemporal_Pivos_ptjpl_CRA\ptjpl_sims_B.csv", index_col='date')
sims.index = pd.to_datetime(sims.index)
sims = sims.dropna()
sims.sort_index(inplace=True)

```

### Visualizando série temporal

```python
# ...gráfico da ET média no pivo selecionado
fig, ax = plt.subplots(figsize=(12,8), dpi=300)

ptjpl['et'].plot(ax=ax, linewidth=1, linestyle=':', marker='o', markersize=6, color='forestgreen', alpha=1)
sims['et'].plot(ax=ax, linewidth=1, linestyle=':', marker='o', markersize=6, color='royalblue', alpha=1)

ax.set_title(title, fontsize=16, fontweight='bold')
ax.set_ylabel('Evapotranspiração (mm/dia)', fontsize=16, fontweight='bold', labelpad=20)
ax.set_xlabel('', fontsize=16, fontweight='bold', labelpad=20)
#ax.set_ylim(0,None)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.legend(labels=['PT-JPL', 'SIMS'])

# Formatar datas no eixo X para DD-MM-YY
formato_data = mdates.DateFormatter('%d-%b-%Y')
ax.xaxis.set_major_formatter(formato_data)
fig.autofmt_xdate()
```
